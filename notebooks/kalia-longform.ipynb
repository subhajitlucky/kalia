# KALIA long-form probe — verify the LAMBADA hypothesis before we build on it

Our corpus has a **median document of 239 tokens** and only **10.8% of documents exceed
1024**. With `context_len = 1024`, almost no training window is a whole document — the
model is trained on fragments stitched across boundaries, and I15 showed it attends across
them. LAMBADA needs the opposite: hold a broad discourse and recall its final word. Hence
**18.4%**, barely above the ~10% a model gets by picking a random capitalised word.

This kernel measures document-length distribution per source so that claim is re-verified
on the real mixture rather than assumed, and confirms a licence-clean long-form source
exists (PG-19: pre-1919 books, out of copyright by construction).

CPU kernel — no GPU quota.
